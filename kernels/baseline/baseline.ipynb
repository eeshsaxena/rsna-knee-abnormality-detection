```python
# RSNA Knee Abnormality Detection - plumbing baseline.
# Emits a valid submission.csv (12 labels, all 0.5) for whatever test set
# is mounted at rerun time. Purpose: validate the code-submission pipeline
# end to end. ROC AUC of a constant column is 0.5 by definition, so this is
# a ~0.5 board anchor, not a modeling attempt.
import glob
import pandas as pd

LABELS = ['ACL', 'MCL', 'Medial Meniscus', 'Lateral Meniscus', 'Medial OA',
          'Lateral OA', 'PF OA', 'Effusion', 'Synovitis', "Baker's",
          'Contusion', 'Fracture']


def find(name):
    hits = sorted(glob.glob(f'/kaggle/input/**/{name}', recursive=True))
    return hits[0] if hits else None


ss = find('sample_submission.csv')
if ss:
    sub = pd.read_csv(ss)
    if 'StudyInstanceUID' not in sub.columns:
        sub = sub.rename(columns={sub.columns[0]: 'StudyInstanceUID'})
else:
    tc = find('test.csv')
    if tc is None:
        raise FileNotFoundError('neither sample_submission.csv nor test.csv found under /kaggle/input')
    sub = pd.read_csv(tc)[['StudyInstanceUID']].copy()

for lbl in LABELS:
    sub[lbl] = 0.5
sub = sub[['StudyInstanceUID'] + LABELS]
sub.to_csv('submission.csv', index=False)
print('submission shape:', sub.shape)
print(sub.head())
```
